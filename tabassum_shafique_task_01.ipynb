           print(7 / 2)  
           output: 3.5

### Part_02: 
           print(7 // 2)  
           output: 3  
Reason: `//` ko *"Floor Division"* kehte hain.
(7 // 2 = 3.5) round about ker ky decimal part hta daita hai or whole no daita hai, isi liye output `3` hai.


### Part_03:

            output: 72 
Reason: "7" aur "2" quotation marks mein hain, isi liye ye yeh dono yha pe numbers nahi balkay strings hain. strings dono ko jor deta hai.

           print(2 + 3 * 4 ** 2)
           output: 50
